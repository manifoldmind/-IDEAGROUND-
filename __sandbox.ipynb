Задача 3: Композиция из нескольких объектов
Условие:
Сложный объект может состоять из нескольких разных компонентов.

Вам нужно:

Создать класс Engine (Двигатель) с методом start(), который возвращает "Двигатель запущен".

Создать класс Wheels (Колеса) с методом rotate(), который возвращает "Колеса вращаются".

Создать класс Car (Машина). В его __init__ создайте два атрибута: self.engine (экземпляр Engine) и self.wheels (экземпляр Wheels).

В классе Car создать метод drive(), который возвращает строку, объединяющую результаты работы компонентов в строго заданном формате: "[результат start()] и [результат rotate()]". Например: "Двигатель запущен и Колеса вращаются"

```python
class Engine:
    def start(self):
        return "Двигатель запущен"

class Wheels:
    def rotate(self):
        return "Колеса вращаются"

class Car:
    def __init__(self):
        self.engine = Engine()
        self.wheels = Wheels()

    def drive(self):
        eng_str = self.engine.start
        wh_str = self.wheels.rotate
        return f"{eng_str()} и {wh_str()}"
    
car01 = Car()
print(car01.drive())
```

Задача 4: Композиция со списком объектов
Условие:
Объект может содержать не один компонент, а целый список.

Вам нужно:

Создать класс Chapter (Глава). Его __init__ должен принимать title, а метод get_title() возвращать его.

Создать класс Book (Книга). Его __init__ должен принимать title и chapters (список строк с названиями глав).

В __init__ класса Book нужно преобразовать список строк chapters в список объектов Chapter и сохранить его в атрибут self.chapters.

Создать в Book метод get_table_of_contents(), который возвращает строку с оглавлением. Первая строка — название книги, далее каждая глава на новой строке в формате "Глава N: [название главы]", где N — порядковый номер, начиная с 1."

```python
class Chapter:
    def __init__(self, title:str):
        self.title = title
    
    def get_title(self): #WARN: get_title - "взять название ГЛАВЫ" (а не книги...)
        return self.title
    
class Book:
    def __init__(self, title, chapters:list[str]):
        self.title = title
        self.chapters = [Chapter(chap) for chap in chapters] #ANALYZED: каж. chap - экземпляр класса Chapter
    
    def get_table_of_contents(self):
        _book_title = self.title
        end_str = "\n"
        chapus = self.chapters
        for chap in chapus:
            end_str += f"Глава {chapus.index(chap)+1}: " + str(chap.get_title()) + "\n"

        ans = _book_title + end_str[:-1]
        return ans
```

```python
# Функция для тестирования
def test_book():
    # Создаём тестовую книгу
    book = Book(
        title="Приключения тестировщика",
        chapters=["Пролог", "Глава первая", "Глава вторая", "Эпилог"]
    )
    # Пытаемся получить оглавление и вывести его

    toc = book.get_table_of_contents()
    print(toc)
    # except Exception as e:
    #     print("Ошибка при вызове get_table_of_contents:")
    #     raise e


# Запуск теста
if __name__ == "__main__":
    test_book()
```
